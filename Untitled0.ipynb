```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Sample input: multiple paragraphs
text = """Natural Language Processing is a branch of Artificial Intelligence.
It helps computers to understand human language.

NLTK is one of the most popular NLP libraries.
It provides tools for tokenization, stemming, tagging and more."""

# Download the punkt resource if not already present
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')


# Split into paragraphs
paragraphs = text.split("\n\n")

# Count paragraphs
num_paragraphs = len(paragraphs)

# Count sentences and words
num_sentences = 0
num_words = 0

for para in paragraphs:
    sentences = sent_tokenize(para)
    num_sentences += len(sentences)
    for sent in sentences:
        words = word_tokenize(sent)
        num_words += len(words)

print("Number of Paragraphs:", num_paragraphs)
print("Number of Sentences:", num_sentences)
print("Number of Words:", num_words)
```
